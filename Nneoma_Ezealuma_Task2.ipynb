```python
Student_Scores = [82, 95, 43, 70, 80, 50, 30, 68, 90]
#using the user defined function

def grade_student(score):
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

#the conditional statement below will Process all scores
for i in range(len(Student_Scores)):
        score = Student_Scores[i]
        grade = grade_student(score)
        status = "PASSED" if score  >= 50 else "FAILED"
        print(f"Student{i+1}:Score = {score}, Grade = {grade} ({status})")
```

```python
#Mini ATM logic in a function

def atm_machine():
    # Step a: Start with a variable balance = 1000
    balance = 1000

    # Step b: use while loop to keep running until Exit
    while True:
        # Step c: Show options
        print("\n--- ATM Menu ---")
        print("1  Check Balance")
        print("2  Deposit Money")
        print("3  Withdraw Money")
        print("4  Exit")

        choice = input("\nEnter choice (1-4): ")

    # Step d: use if_elif_else for each action(option selected)
        if choice == "1":
            # Check balance
            print(f"Current balance: ${balance}")

        elif choice == "2":
           # Deposit money
           amount = float(input("Enter deposit amount: $"))
           balance += amount # Add money
           print(f"Deposited ${amount}. New balance: ${balance}")

        elif choice == "3":
           # withdraw money
           amount = float(input("Enter withdrawal amount: $"))
           if amount <= balance: # Only if enough balance
              balance -= amount # Subtract
              print(f"Withdraw ${amount}. New balance: ${balance}")
              
           else:
            print("Insufficient funds!")

        elif choice == "4":
          # Exit
          print("Thank you for using ATM. Goodbye!")
          break #Exit the while loop

        else:
         print("Invalid choice! Please enter 1-4.")

# Call the function which you defined from the onset to run the ATM
atm_machine()

```
